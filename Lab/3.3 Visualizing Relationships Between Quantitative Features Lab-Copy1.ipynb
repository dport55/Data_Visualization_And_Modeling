```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Employee_productivity.csv')
df.head()
```

```python
sns.scatterplot(x='YearsAtCompany',y='ProductivityScore', data=df)
plt.title('Employee Productivity by Years at the Company')
plt.ylabel('Productivity Score')
plt.show()
```

```python
import statsmodels.api as sm

X = sm.add_constant(df['YearsAtCompany'])
y = df['ProductivityScore']

model = sm.OLS(y,X).fit()

model.summary()
```

```python
sns.scatterplot(x='HoursWorked',y='ProductivityScore', data=df)
plt.title('Employee Productivity by Hours Worked')
plt.ylabel('Productivity Score')
plt.show()
```

```python
X = sm.add_constant(df['HoursWorked'])
y = df['ProductivityScore']

model = sm.OLS(y,X).fit()

model.summary()
```

```python
sns.scatterplot(x='FocusScore',y='ProductivityScore', data=df)
plt.title('Employee Productivity by Focus Score')
plt.ylabel('Productivity Score')
plt.show()
```

```python
X = sm.add_constant(df['FocusScore'])
y = df['ProductivityScore']

model = sm.OLS(y,X).fit()

model.summary()
```

```python

df.loc[(df['Department'] == 'HR'), 'Dep_cat'] = 'HR'
df.loc[(df['Department'] == 'Engineering'), 'Dep_cat'] = 'Engineering'
df.loc[(df['Department'] == 'Marketing'), 'Dep_cat'] = 'Marketing'
df.loc[(df['Department'] == 'Sales'), 'Dep_cat']= 'Sales'



sns.scatterplot(x='FocusScore',y='ProductivityScore', hue='Dep_cat', data=df)
plt.title('Employee Productivity by Department')
plt.ylabel('Productivity Score')
plt.show()
```

```python

df.loc[df['Department'] == 'HR', 'De_cat'] = 0
df.loc[df['Department'] == 'Engineering', 'De_cat'] = 1
df.loc[df['Department'] == 'Marketing', 'De_cat'] = 2
df.loc[df['Department'] == 'Sales', 'De_cat'] = 3

X = sm.add_constant(df[['FocusScore', 'De_cat']]) 
y = df['ProductivityScore']

model = sm.OLS(y,X).fit() 

model.summary()
```
